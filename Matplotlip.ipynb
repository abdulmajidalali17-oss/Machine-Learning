```python
import numpy as np
import matplotlib.pyplot as plt

# Create NumPy arrays
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1, 4, 9, 16, 25])

# Plot the data
plt.plot(X, Y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("X vs Y")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 3, 7, 6])

plt.scatter(X, Y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Scatter Plot")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.linspace(0, 10, 100)

Y_exp = np.exp(X)
Y_sin = np.sin(X)

plt.plot(X, Y_exp, label="Exponential")
plt.plot(X, Y_sin, label="Sine")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Exponential and Sine Functions")
plt.legend()

plt.show()
```

```python
nut_butter_prices = {
    "Almond butter": 10,
    "Peanut butter": 8,
    "Cashew butter": 12
}

fig, ax = plt.subplots()

ax.bar(
    nut_butter_prices.keys(),
    nut_butter_prices.values()
)

ax.set(
    title="Dan's Nut Butter Store",
    ylabel="Price ($)"
)
```

```python
fig, ax = plt.subplots()

ax.barh(
    nut_butter_prices.keys(),
    nut_butter_prices.values()
)

ax.set(
    title="Dan's Nut Butter Store",
    xlabel="Price ($)"
)
```
