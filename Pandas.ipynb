```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmad", "Omar", "Sara", "Maya"],
    "Age": [22, 25, 21, 24, 23],
    "Salary": [800, 1200, 700, 1500, 1100],
    "Department": ["IT", "AI", "IT", "HR", "AI"]
}

df = pd.DataFrame(data)

df.to_csv("employees.csv", index=False)

print("CSV file created!")
```
